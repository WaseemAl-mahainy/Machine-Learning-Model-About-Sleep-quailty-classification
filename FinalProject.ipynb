## preprocessing 

```python
# we have here the first col is unique , personID so we drop that column
```

```python
df.drop(["Person ID"] , axis=1 , inplace=True)
df
```

```python
## we saw here the gender col it's object not float we cant train the modle like that we need to handle this by using label encoder
```

```python
df["Gender"].unique()
```

```python
from sklearn.preprocessing import LabelEncoder
L=LabelEncoder()
df["Gender"]=L.fit_transform(df["Gender"])
df
```

```python
## we can see here the sleep disorder col has NaN valuse so we need to handle this problem then we go to next step
```

```python
df["Sleep Disorder"].isnull().sum() # we have here 219 Nan value and our data has 374 row so we cant delete it we need to fix it by using imputer 
```

```python
# we can apply the imputer here before the spliting becasuse it's String value
from sklearn.impute import SimpleImputer
s=SimpleImputer(missing_values=np.nan, strategy="constant" , fill_value="none")
s.fit(df[["Sleep Disorder"]])
df["Sleep Disorder"]=s.transform(df[["Sleep Disorder"]]).ravel()
```

#### drowing the boxplot for every feature before handline

```python
# Identify numerical columns
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

# Step 1: Draw boxplots before preprocessing
print("Boxplots BEFORE Preprocessing:")
for column in numerical_columns:
    plt.figure(figsize=(7, 2))
    sns.boxplot(data=df ,x=column,  color='lightblue')
    plt.title(f'Boxplot for {column} (Before Preprocessing)', fontsize=14)
    plt.show()
```

```python
## now we need to convert all the object values to float without bais so we use one hot encoder
```

```python
df["BMI Category"].unique()
```

```python
df["Occupation"].unique()
```

```python
df=pd.get_dummies(df,columns=["Occupation" , "BMI Category", "Sleep Disorder"] , dtype=int)
df
```

```python
# last thing we have the pressure here we need to handle this problem we need to create a custom extract function 
```

```python
def extract_high_pre(z):
    return int(z.split("/")[0])
extract_high_pre("140/95")
```

```python
df["High Blood Pressure"]=df["Blood Pressure"].apply(extract_high_pre)
```

```python
def extract_low_pre(s):
    return int (s.split("/")[1])
extract_low_pre("140/95")
```

```python
df["Low Blood Pressure"]=df["Blood Pressure"].apply(extract_low_pre)
```

```python
df
```

```python
# then we drop the Blood Pressure col to be ready 
```

```python
df.drop(["Blood Pressure"] , axis=1 , inplace=True)
df
```

#### we saw here there is an outlier on heart Rate we remove it

```python
print("Removing Outliers...")
for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    # Remove rows with outliers
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    if not outliers.empty:
        print(f"Removing {len(outliers)} outliers in '{column}'...")
        df = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
```

```python
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import label_binarize
import matplotlib.pyplot as plt
import numpy as np

# Binarize the labels for multiclass classification
y_test_binarized = label_binarize(y_test, classes=np.unique(y_test))  
n_classes = y_test_binarized.shape[1]

# here we get the scores : predict_proba or decision_function 
if hasattr(clf2_5, "predict_proba"):
    y_scores = clf2_5.predict_proba(X_test)
elif hasattr(clf2_5, "decision_function"):
    y_scores = clf2_5.decision_function(X_test)
else:
    raise ValueError("Model must have predict_proba or decision_function.")

# compute ROC and AUC for each class
fpr = {}
tpr = {}
roc_auc = {}

for i in range(n_classes):
    fpr[i], tpr[i], _ = roc_curve(y_test_binarized[:, i], y_scores[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])

# compute micro-average ROC and AUC
fpr["micro"], tpr["micro"], _ = roc_curve(y_test_binarized.ravel(), y_scores.ravel())
roc_auc["micro"] = auc(fpr["micro"], tpr["micro"])

# Plot ROC curves for each class
plt.figure(figsize=(10, 8))
colors = plt.cm.rainbow(np.linspace(0, 1, n_classes))
for i, color in zip(range(n_classes), colors):
    plt.plot(fpr[i], tpr[i], color=color, lw=2,
             label=f'Class {i} (AUC = {roc_auc[i]:.2f})')

# Plot micro-average ROC curve
plt.plot(fpr["micro"], tpr["micro"], color='black', linestyle='--', lw=2,
         label=f'Micro-average (AUC = {roc_auc["micro"]:.2f})')

# Diagonal line
plt.plot([0, 1], [0, 1], color='red', linestyle='--')

# Plot settings
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve for Multiclass Classification')
plt.legend(loc='lower right')
plt.grid(alpha=0.3)
plt.show()
```
